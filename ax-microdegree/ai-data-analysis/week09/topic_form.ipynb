```python
import pandas as pd

DATA = 'https://raw.githubusercontent.com/Daeil-Noh/Lectures/main/ax-microdegree/ai-data-analysis/data/'
my_file = DATA + 'process.csv'         # ← 내 데이터 주소나 파일 이름으로 바꾸세요

df = pd.read_csv(my_file)
print('크기:', df.shape[0], '행 ×', df.shape[1], '열')
print('열 이름:', list(df.columns))
print('빈 칸이 있는 열:', df.columns[df.isna().any()].tolist())
```
